```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
currency = pd.read_csv("../data/currency.csv")
currency.head()
```

```python
currency.columns = ["date", "value"]
currency["date"] = pd.to_datetime(currency["date"])
currency.set_index("date", inplace=True)
```

```python
plt.figure(figsize=(10, 6))
currency["value"].plot()
plt.grid()
plt.xlabel("date")
plt.ylabel("US Dollar")
plt.title("Canadian Dollars to U.S. Dollar Spot Exchange Rate")
plt.show()
```
